# Parquet 文件数据量统计

统计 `data` 目录下每个 parquet 文件的行数、列数、文件大小等信息。

```python
import os
import glob
import pandas as pd
from pathlib import Path

# 获取 data 目录下所有 parquet 文件
data_dir = Path(".")  # notebook 位于 data 目录下
parquet_files = sorted(data_dir.glob("../output/onpolicy_distillation.parquet"))

print(f"找到 {len(parquet_files)} 个 parquet 文件:\n")
for f in parquet_files:
    print(f"  {f}")
```

```python
# 输出每个文件的前10个样本查看基本数据
for fp in parquet_files:
    df = pd.read_parquet(fp)
    print(f"{'='*80}")
    print(f"文件: {fp} (共 {len(df)} 行)")
    print(f"列名: {df.columns.tolist()}")
    print(f"{'='*80}")
    
    # 显示前10个样本
    sample_count = min(5, len(df))
    for i in range(sample_count):
        row = df.iloc[i]
        print(f"--- 样本 {i+1} ---")
        for col in df.columns:
            val = row[col]
            val_str = str(val)
            print(f"  {col}: {val_str}")
```

```python
# 统计每个 parquet 文件的信息
stats = []

for fp in parquet_files:
    file_size_mb = os.path.getsize(fp) / (1024 * 1024)  # MB
    df = pd.read_parquet(fp)
    tmp = {
        "文件名": str(fp),
        "行数": len(df),
        "列数": len(df.columns),
        "列名": ", ".join(df.columns.tolist()),
        "文件大小(MB)": round(file_size_mb, 2),
    }
    print(tmp)
    stats.append(tmp)

stats_df = pd.DataFrame(stats)
stats_df
```

```python
# 输出每个文件的前10个样本查看基本数据
for fp in parquet_files:
    df = pd.read_parquet(fp)
    print(f"{'='*80}")
    print(f"文件: {fp} (共 {len(df)} 行)")
    print(f"列名: {df.columns.tolist()}")
    print(f"{'='*80}")
    
    # 显示前10个样本
    sample_count = min(1, len(df))
    for i in range(sample_count):
        row = df.iloc[i]
        print(f"--- 样本 {i+1} ---")
        for col in df.columns:
            val = row[col]
            val_str = str(val)
            print(f"  {col}: {val_str}")

```

```python
# 汇总统计
total_rows = stats_df["行数"].sum()
total_size_mb = stats_df["文件大小(MB)"].sum()

print("=" * 60)
print(f"总文件数: {len(parquet_files)}")
print(f"总行数: {total_rows:,}")
print(f"总文件大小: {total_size_mb:.2f} MB ({total_size_mb / 1024:.2f} GB)")
print("=" * 60)
print()
print("各文件行数占比:")
for _, row in stats_df.iterrows():
    pct = row["行数"] / total_rows * 100 if total_rows > 0 else 0
    print(f"  {row['文件名']}: {row['行数']:,} 行 ({pct:.1f}%)")
```

```python
# 可视化 - 各文件行数柱状图
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# 行数柱状图
short_names = [Path(r["文件名"]).stem for _, r in stats_df.iterrows()]
axes[0].barh(short_names, stats_df["行数"].values)
axes[0].set_xlabel("行数")
axes[0].set_title("各 Parquet 文件行数")
for i, v in enumerate(stats_df["行数"].values):
    axes[0].text(v, i, f" {v:,}", va="center", fontsize=9)

# 文件大小柱状图
axes[1].barh(short_names, stats_df["文件大小(MB)"].values)
axes[1].set_xlabel("文件大小 (MB)")
axes[1].set_title("各 Parquet 文件大小")
for i, v in enumerate(stats_df["文件大小(MB)"].values):
    axes[1].text(v, i, f" {v:.2f}", va="center", fontsize=9)

plt.tight_layout()
plt.show()
```
